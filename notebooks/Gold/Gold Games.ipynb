```python
import json
from pyspark.sql import Row
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

# Dim Games

```python
dim_games = (
    spark.table("silver.mlb_schedule")
    .select("game_pk", "game_date", "home_team_id", "away_team_id", "game_status", "double_header")
    .withColumn("gold_processed_at", current_timestamp())
)
```

```python
dim_games.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold.dim_games")
```

```python
non_final = dim_games.filter(col("game_status") != "Final")
print(f"Non-final games (postponed/suspended/etc): {non_final.count()}")
display(non_final)
```
