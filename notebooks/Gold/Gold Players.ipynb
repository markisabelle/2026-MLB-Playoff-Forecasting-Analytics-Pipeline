```python
import json
from pyspark.sql import Row
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

# Dim Players

```python
batting = spark.table("silver.mlb_batting_stats")
pitching = spark.table("silver.mlb_pitching_stats")

player_identities = (
    batting.select("player_id", "player_name")
    .union(pitching.select("player_id", "player_name"))
    .dropDuplicates(["player_id"])
)
```

```python
dim_players = player_identities.withColumn("gold_processed_at", current_timestamp())

dim_players.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold.dim_players")
```
