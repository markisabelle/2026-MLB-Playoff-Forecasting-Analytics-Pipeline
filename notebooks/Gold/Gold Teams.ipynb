```python
dim_teams = (
    spark.table("silver.mlb_teams")
    .withColumn("gold_processed_at", current_timestamp())
)

dim_teams.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold.dim_teams")

print(f"dim_teams rows: {dim_teams.count()}")
display(dim_teams.limit(5))
```
